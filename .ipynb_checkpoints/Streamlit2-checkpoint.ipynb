### WS(로또 생성기)

```python
%%writefile test.py
import streamlit as st
import random
import datetime

st.title(':sparkles:로또 생성기:sparkles:')


def generate_lotto():
    lotto = set()

    while len(lotto) < 6:
        number = random.randint(1, 46)
        lotto.add(number)

    lotto = list(lotto)
    lotto.sort()
    return lotto

# st.subheader(f'행운의 번호: :green[{generate_lotto()}]')
# st.write(f"생성된 시각: {datetime.datetime.now().strftime('%Y-%m-%d %H:%M')}")

button = st.button('로또를 생성해 주세요!')

if button:
    for i in range(1, 6):
        st.subheader(f'{i}. 행운의 번호: :green[{generate_lotto()}]')
    st.write(f"생성된 시각: {datetime.datetime.now().strftime('%Y-%m-%d %H:%M')}")
```

```python
%%writefile test.py

import streamlit as st
import pandas as pd
import numpy as np
from datetime import datetime


st.title('Streamlit 01. Data 출력')



with st.echo():
    # DataFrame 생성
    dataframe = pd.DataFrame({
        'first column': [1, 2, 3, 4],
        'second column': [10, 20, 30, 40],
        'third column': ['bar', 'foo']*2,
        'fourth column': [True, False, True, False],
        'fifth column': ['📊 Data Exploration', '📈 Data Visualization','🤖 LLM','📊 Data Exploration'],
        'sixth column': pd.date_range('2023/07/11', periods=4, freq='30T'),
        'seventh column': [[1,2,3,4], [10, 20, 30, 40], [100, 200, 300, 400], [1000, 2000, 3000, 4000]],
        'eighth column': ['https://www.google.com', 'https://www.naver.com', 'https://stackoverflow.com', 'https://streamlit.io'],
        'ninth column': [
            'https://pypi.org/static/images/logo-small.2a411bc6.svg',
            'https://pypi.org/static/images/logo-large.9f732b5f.svg',
            'https://pypi.org/static/images/logo-small.2a411bc6.svg',
            'https://pypi.org/static/images/logo-large.9f732b5f.svg']
    })

with st.echo():
    # DataFrame
    # use_container_width 기능은 데이터프레임을 컨테이너 크기에 확장할 때 사용합니다. (True/False)
    st.dataframe(dataframe, use_container_width=False)
    #st.dataframe(dataframe.style.highlight_min(axis=0), use_container_width=False)

with st.echo():
    # 테이블(static)
    # DataFrame과는 다르게 interactive 한 UI 를 제공하지 않습니다.
    st.table(dataframe.style.highlight_max(axis=0))

with st.echo():
    # DataFrame
    # use_container_width 기능은 데이터프레임을 컨테이너 크기에 확장할 때 사용합니다. (True/False)
    edited = st.data_editor(dataframe)

    st.write(edited.iloc[0, 0])

with st.echo():
    # Column Config: Column
    st.dataframe(
        dataframe,
        column_config={
            'third column': st.column_config.Column(
                'String Column',
                help='문자열로 구성된 열입니다.',
                width='medium'
            )
        }
    )

with st.echo():
    # Column Config: Text Column
    st.data_editor(
        dataframe,
        column_config={
            'third column': st.column_config.TextColumn(
                'String Column',
                help='문자열로 구성된 열입니다.',
                max_chars=20,
                validate='[a-z]'
            )
        }
    )

with st.echo():
    # Column Config: Number Column
    st.data_editor(
        dataframe,
        column_config={
            'second column': st.column_config.NumberColumn(
                'Number Column',
                help='숫자로 구성된 열입니다..',
                min_value=0,
                max_value=1000,
                step=1,
                format='$%d'
            )
        }
    )

with st.echo():
    # Column Config: Checkbox Column
    st.data_editor(
        dataframe,
        column_config={
            'fourth column': st.column_config.CheckboxColumn(
                'Checkbox Column',
                help='체크상자로 구성된 열입니다..',
                default=False
            )
        }
    )

with st.echo():
    # Column Config: Selectbox Column
    st.data_editor(
        dataframe,
        column_config={
            'fifth column': st.column_config.SelectboxColumn(
                'Selectbox Column',
                help='선택상자로 구성된 열입니다..',
                options=[
                    '📊 Data Exploration',
                    '📈 Data Visualization',
                    '🤖 LLM'
                ]
            )
        }
    )

with st.echo():
    # Column Config: Datetime Column (step=1s)
    # Column Config: Date Column (step=1day)
    # Column Config: time Column (step=1s)
    st.data_editor(
        dataframe,
        column_config={
            'sixth column': st.column_config.DatetimeColumn(
                'Datetime Column',
                help='날짜로 구성된 열입니다..',
                min_value = datetime(2023, 7, 11),
                max_value = datetime(2023, 7, 12),
                format='YYYY년 MM월 DD일 HH시 mm분 a',
                step=1
            )
        }
    )

with st.echo():
    # Column Config: List Column
    st.data_editor(
        dataframe,
        column_config={
            'seventh column': st.column_config.ListColumn(
                'List Column',
                help='리스트로 구성된 열입니다.',
            )
        }
    )

with st.echo():
    # Column Config: Link Column
    st.data_editor(
        dataframe,
        column_config={
            'eighth column': st.column_config.LinkColumn(
                'Link Column',
                help='링크로 구성된 열입니다.',
            )
        }
    )

with st.echo():
    # Column Config: Image Column
    st.data_editor(
        dataframe,
        column_config={
            'ninth column': st.column_config.ImageColumn(
                'Image Column',
                help='이미지 링크로 구성된 열입니다.',
            )
        }
    )

with st.echo():
    # Column Config: Line Chart Column
    st.data_editor(
        dataframe,
        column_config={
            'seventh column': st.column_config.LineChartColumn(
                'Line Chart Column',
                help='선 차트로 구성된 열입니다.',
                y_min=0,
                y_max=500
            )
        }
    )

with st.echo():
    # Column Config: Bar Chart Column
    st.data_editor(
        dataframe,
        column_config={
            'seventh column': st.column_config.BarChartColumn(
                'Bar Chart Column',
                help='Bar 차트로 구성된 열입니다.',
                y_min=0,
                y_max=500
            )
        }
    )

with st.echo():
    # Column Config: Progress Column
    st.data_editor(
        dataframe,
        column_config={
            'second column': st.column_config.ProgressColumn(
                'Progress Column',
                help='프로그레스 바로 구성된 열입니다.',
                min_value=0,
                max_value=100,
                format='$%d'
            )
        }
    )

with st.echo():
    # 메트릭
    # delta_color = 'normal', 'inverse' or 'off'
    # label_visibility = 'visible', 'hidden' or 'collapsed'
    st.metric(label="온도", value="10°C", delta="1.2°C", help='기본')
    st.metric(label="GS", value="36,250 원", delta="1,150 원 (+3.28%)", delta_color='inverse', help='델타 색상 반전')

with st.echo():
    # 컬럼으로 영역을 나누어 표기한 경우
    col1, col2, col3 = st.columns(3)
    col1.metric(label="달러USD", value="1,228 원", delta="-12.00 원", label_visibility='visible')
    col2.metric(label="일본JPY(100엔)", value="958.63 원", delta="-7.44 원", label_visibility='hidden')
    col3.metric(label="유럽연합EUR", value="1,335.82 원", delta="11.44 원", label_visibility='collapsed')

with st.echo():
    # st.json
    st.json({
        'foo': 'bar',
        'baz': 'boz',
        'stuff': [
            'stuff 1',
            'stuff 2',
            'stuff 3',
            'stuff 5',
        ],
    }, expanded=True)
```

### Sesstion State

```python
%%writefile test.py

import streamlit as st

st.title('State Management')

# session state
st.header('Session States')


def first_page():

    st.title('첫 페이지 입니다.')
    st.write('UI 컴포넌트 상태를 관리합니다.')

    checkbox_key = 1234
    cur_state = False
    check = 'current_state'

    if check in st.session_state:
        cur_state = st.session_state[check]

    agree = st.checkbox('오늘 수업 종료할까요?', cur_state, key=checkbox_key)


    if agree:
        msg = '네 ~~ 야외 수업해요 :100:'
        st.write(msg)
        st.session_state['current_state'] = True
        st.session_state['msg'] = msg
        print('agree true 수행 완료')

    else:
        st.session_state['current_state'] = False


def second_page():

        # 세션에 키 등록
    if 'button_state' not in st.session_state:
        st.session_state['button_state'] = False


    # 버튼 클릭
    left, center, right = st.columns(3)
    button = center.button('버튼을 눌러보세요')

    if button:
        msg = '버튼이 눌렸습니다.'
        st.write(msg)
        st.session_state['button_state'] = True
        st.session_state['button_write'] = msg

    else:
        if st.session_state['button_state'] == True:
            st.write(st.session_state['button_write'])

    # 체크 박스
    agree1 = st.checkbox('동의 하십니까?')

    if agree1:
        st.write('동의 해주셔서 감사합니다 :100:')

with st.sidebar:
    menu = st.selectbox(
        '페이지를 선택하세요.',
        ('첫화면', '두번째화면'),
        index=0
    )
if menu == '첫화면':
    first_page()
else:
    second_page()
```

```python
%%writefile test.py
import streamlit as st

# 필요 패키지 추가
import pandas as pd
import numpy as np
import matplotlib
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
import seaborn as sns

# 실습 소개 페이지 출력 함수
# 소개 페이지는 기본으로 제공됩니다.
def front_page():
    st.title('데이터 분석 대시보드 실습')
    st.write('이 실습은 데이터 분석 대시보드를 생성하는 실습입니다.')
    st.markdown('''
        - 사이드 바 추가
          - selectmenu 이용
        - 처음 화면 출력
          - 제공된 front_page 함수 출력
        - EDA 페이지 출력
          - 파일 업로드
          - df 읽기
          - 변수 분포 시각화 (expander)
            - 1개 선택
          - 수치형 변수 관계 시각화 (expander)
            - 선택 목록에 수치형만 표시
            - 수치형 2개 선택
        - 사이드 바의 메뉴를 변경한 후에도 EDA 결과들이 유지되도록 session state를 설정해보세요.
    ''')

def eda_page():
    st.title('EDA 페이지')

    file = st.file_uploader("파일 선택")

    if file is not None:
        st.session_state['current_variable'] = None
        st.session_state['current_variables'] = None
        df = pd.read_csv(file)
        st.session_state['dataframe'] = df

    if st.session_state.get('dataframe') is not None:
        df = st.session_state['dataframe']
        st.write(df)
        with st.expander('변수 분포 시각화', expanded=True):
            if st.session_state['current_variable'] is not None:
                index, _ = st.session_state['current_variable']
                index = (['변수 선택']+list(df.columns)).index(index)
            else:
                index = 0

            col = st.selectbox('변수를 선택하세요.', ['변수 선택']+list(df.columns), index=index)

            if col != '변수 선택':
                if df[col].dtype in ['int64', 'float64']:
                    plt.figure()
                    ax = df[col].hist(bins=30)
                    plt.title(col)
                    st.session_state['current_variable'] = [col, ax.get_figure()]
                else:
                    plt.figure()
                    ax = df[col].value_counts().sort_index().plot.bar()
                    plt.title(col)
                    st.session_state['current_variable'] = [col, ax.get_figure()]

            if st.session_state['current_variable'] is not None:
                st.pyplot(st.session_state['current_variable'][1])


        with st.expander('연속형 변수 관계 시각화', expanded=True):
            if st.session_state['current_variables'] is not None:
                index, _ = st.session_state['current_variables']
            else:
                index = []
            cols = df.select_dtypes(['int64', 'float64']).columns
            selected_cols = st.multiselect(
                '변수 선택',
                cols,
                index,
                max_selections=2)

            if len(selected_cols) == 2:
                plt.figure()
                plot = sns.scatterplot(x=selected_cols[0], y=selected_cols[1], data=df)
                st.session_state['current_variables'] = [selected_cols, plot.get_figure()]

            if st.session_state['current_variables'] is not None:
                st.pyplot(st.session_state['current_variables'][1])

# sidebar 추가
with st.sidebar:
    menu = st.selectbox(
        '메뉴를 선택하세요.',
        ('처음 화면', 'EDA 페이지'),
        index=0
    )
if menu == '처음 화면':
    front_page()
else:
    eda_page()
```
